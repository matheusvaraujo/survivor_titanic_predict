```python
import pandas as pd
```

```python
data_train = pd.read_csv('/home/lactea/Documentos/survivor_titanic_predict/data_titanic/titanic/train.csv')
data_test = pd.read_csv('/home/lactea/Documentos/survivor_titanic_predict/data_titanic/titanic/test.csv')
```

```python
data_train.rename(columns={
    'PassengerId': 'ID_Passageiro',
    'Survived': 'Sobreviveu',
    'Pclass': 'Classe',
    'Name': 'Nome',
    'Sex': 'Sexo',
    'Age': 'Idade',
    'SibSp': 'Irmaos_Conjuges',
    'Parch': 'Pais_Filhos',
    'Ticket': 'Bilhete',
    'Fare': 'Tarifa',
    'Cabin': 'Cabine',
    'Embarked': 'Embarque'
}, inplace = True)
```

```python
data_test.rename(columns={
    'PassengerId': 'ID_Passageiro',
    'Pclass': 'Classe',
    'Name': 'Nome',
    'Sex': 'Sexo',
    'Age': 'Idade',
    'SibSp': 'Irmaos_Conjuges',
    'Parch': 'Pais_Filhos',
    'Ticket': 'Bilhete',
    'Fare': 'Tarifa',
    'Cabin': 'Cabine',
    'Embarked': 'Embarque'
}, inplace = True)
```

```python
data_train
```

```python
data_train = data_train.drop(columns=['ID_Passageiro', 'Nome', 'Bilhete', 'Cabine'])
data_test = data_test.drop(columns=['ID_Passageiro', 'Nome', 'Bilhete', 'Cabine'])
```

```python
data_train
```

```python
medianas = data_train.groupby('Classe')['Idade'].median()

print(medianas)
```

```python
data_train['Idade'] = data_train['Idade'].fillna(
    data_train['Classe'].map(medianas)
)
```

```python
data_test['Idade'] = data_test['Idade'].fillna(
    data_test['Classe'].map(medianas)
)
```

```python
data_train.dropna(inplace=True)
data_train.isnull().sum()
```

```python
data_train['Idade'] = data_train['Idade'].astype(int)
data_test['Idade'] = data_test['Idade'].astype(int)
```

```python
data_train['qtd_familiares'] = data_train['Irmaos_Conjuges'] + data_train['Pais_Filhos']
data_test['qtd_familiares'] = data_test['Irmaos_Conjuges'] + data_test['Pais_Filhos']
```

```python
data_train_encoded = pd.get_dummies(
    data_train,
    columns=['Classe', 'Sexo', 'Embarque'],
    dtype=int
)
```

```python
data_test_encoded = pd.get_dummies(
    data_test,
    columns=['Classe', 'Sexo', 'Embarque'],
    dtype=int
)
```

```python
data_train_encoded
```

```python
data_test_encoded
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = data_train_encoded.drop(columns=['Sobreviveu'])
y = data_train_encoded['Sobreviveu']
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.1,
    random_state=7,
    stratify=y
)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt = DecisionTreeClassifier(
    max_depth=10,
    random_state=7
)

dt.fit(x_train, y_train)
```

```python
dt_model = dt.predict(x_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

```python
print('Acurácia', accuracy_score(y_test, dt_model))
print('Precisão', precision_score(y_test, dt_model))
print('Recall', recall_score(y_test, dt_model))
print('F1-Score', f1_score(y_test, dt_model))
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    random_state=7
)

rf.fit(x_train, y_train)
```

```python
rf_model = rf.predict(x_test)
```

```python
print('Acurácia', accuracy_score(y_test, rf_model))
print('Precisão', precision_score(y_test, rf_model))
print('Recall', recall_score(y_test, rf_model))
print('F1-Score', f1_score(y_test, rf_model))
```

```python
from xgboost import XGBClassifier
```

```python
xgb = XGBClassifier(
    n_estimators=200,
    max_depth=20,
    random_state=7,
)

xgb.fit(x_train, y_train)
```

```python
xgb_model = xgb.predict(x_test)
```

```python
print('Acurácia', accuracy_score(y_test, xgb_model))
print('Precisão', precision_score(y_test, xgb_model))
print('Recall', recall_score(y_test, xgb_model))
print('F1-Score', f1_score(y_test, xgb_model))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression(
    max_iter=1000,
    random_state=7
)

lr.fit(x_train, y_train)
```

```python
lr_model = lr.predict(x_test)
```

```python
print('Acurácia', accuracy_score(y_test, lr_model))
print('Precisão', precision_score(y_test, lr_model))
print('Recall', recall_score(y_test, lr_model))
print('F1-Score', f1_score(y_test, lr_model))
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
knn = KNeighborsClassifier(
    n_neighbors=5,
    metric='euclidean',
)

knn.fit(x_train_scaled, y_train)
```

```python
knn_model = knn.predict(x_test_scaled)
```

```python
print('Acurácia', accuracy_score(y_test, knn_model))
print('Precisão', precision_score(y_test, knn_model))
print('Recall', recall_score(y_test, knn_model))
print('F1-Score', f1_score(y_test, knn_model))
```

```python
from sklearn.ensemble import VotingClassifier
```

```python
voting = VotingClassifier(
    estimators=[
        ('decision_tree', dt),
        ('random_forest', rf),
        ('XGBoost', xgb),
        ('logistic_regression', lr)
    ],
    voting='soft',
    weights=[4,3,5,4]
)

voting.fit(x_train, y_train)
```

```python
voting_model = voting.predict(x_test)
```

```python
print('Acurácia', accuracy_score(y_test, voting_model))
print('Precisão', precision_score(y_test, voting_model))
print('Recall', recall_score(y_test, voting_model))
print('F1-Score', f1_score(y_test, voting_model))
```

```python
from sklearn.ensemble import StackingClassifier
```

```python
stacking = StackingClassifier(
    estimators=[
        ('decision_tree', dt),
        ('random_forest', rf),
        ('XGBoost', xgb),
        ('logistic_regression', lr)
    ],

    final_estimator=XGBClassifier(
        n_estimators=200,
        max_depth=10,
        random_state=7
    ),

    cv=5,

    stack_method='predict_proba',

    passthrough=False
)
```

```python
stacking.fit(x_train, y_train)
stacking_model = stacking.predict(x_test)
```

```python
print('Acurácia', accuracy_score(y_test, stacking_model))
print('Precisão', precision_score(y_test, stacking_model))
print('Recall', recall_score(y_test, stacking_model))
print('F1-Score', f1_score(y_test, stacking_model))
```
